```python
import numpy as np
import pandas as pd
from pandas import DataFrame
from sklearn.linear_model import LinearRegression
```

```python
df = DataFrame({ "data": [1, 2, 3, 4, 5, 6], "value":[2,4,6,8,10,12] })
X = df[["data"]]
y = df[["value"]]
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
res = model.predict([[7]])
res
```
